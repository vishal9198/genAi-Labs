```python
from google.colab import userdata
import os

os.environ["GROQ_API_KEY"]= userdata.get('groqApiKey')
from langchain.chat_models import init_chat_model
from langchain.messages import HumanMessage, AIMessage, SystemMessage

model = init_chat_model("llama-3.1-8b-instant",model_provider="groq")
system_msg = SystemMessage("You are a helpful assistant.")
human_msg = HumanMessage("Hello, how are you?")

# Use with chat models
messages = [system_msg, human_msg]
response = model.invoke(messages)  #
print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate
travel_template= "suggest me a place for visit in {country}"
prompt_template= ChatPromptTemplate.from_messages([("system",travel_template),("user","{country}")])

prompt= prompt_template.invoke({"country": "france"})

model.invoke(prompt)
```

```python
prompt_template= ChatPromptTemplate.from_template("Suggest me name of a place to visit in a {country} in the month of {month} in a single word")
prompt= prompt_template.invoke({"country": "france","month":"June"})

model.invoke(prompt)
```

chaining


```python
from langchain_core.output_parsers import StrOutputParser
chain = prompt_template | model | StrOutputParser()
chain.invoke({"country": "france","month":"June"})
```

```python
# Install dependencies
!pip install -q langgraph langchain

# Import Graph
from langgraph.graph import Graph

# First function
def func1(inp):
    x = inp["x"]
    y = inp["y"]

    return x + y

# Second function
def func2(inp):
    return 2 * inp

# Create workflow graph
workflow = Graph()

# Add nodes
workflow.add_node("node_1", func1)
workflow.add_node("node_2", func2)

# Connect nodes
workflow.add_edge("node_1", "node_2")

# Define start and end
workflow.set_entry_point("node_1")
workflow.set_finish_point("node_2")

# Compile graph
app = workflow.compile()

# Run graph
result = app.invoke({
    "x": 10,
    "y": 5
})

print(result)
```
